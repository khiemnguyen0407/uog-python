# Variables

## Why do we need variables?

Literals that we never use again are useless. Solving complicated problems with a computer is only possible if we can **store** data and intermediate results, and **retrieve** them later. Without a way of storing values, we could not even use a pocket calculator! During a computation, data are kept in the computer's memory, usually the *random-access memory* (RAM). Every programming language therefore needs a mechanism for saving values in memory and accessing them whenever they are needed. This is what **variables** are for.

You can think of **a variable as a name that refers to a value stored somewhere in memory**. Through this name, we can read the value, replace it with a new one, or delete the name altogether when we no longer need it.

> **💡 You already know variables**
>
> You have been working with variables since secondary school. When you solve the linear equation $a x + b = 0$ for the unknown $x$, the letters $a$ and $b$ hold actual numbers, and the solution is "stored" in $x$. Similarly, when you solve the quadratic equation $a_2 x^2 + a_1 x + a_0 = 0$ for $x$, the coefficients $a_0$, $a_1$ and $a_2$ hold actual values.

## Variables in a nutshell

- A variable is a **name** (an alias) that refers to a value stored in memory.
- The actual location in physical memory is chosen by Python, not by you, and it may change from one run to the next.
- In CPython (the standard implementation of Python), the built-in function `id()` returns a number that identifies the object a name refers to — in practice, its memory address.

![Variables are names that refer to objects. After y = x, both x and y refer to the same object, whereas z refers to a different one.](figures/variables-and-objects.svg)

**Figure:** Variables are names that refer to objects. After `y = x`, both `x` and `y` refer to the *same* object, whereas `z` refers to a different one.

## Naming rules and conventions

**Rules** (enforced by Python):

- Names may contain letters, digits and the underscore `_`.
- The first character must be a letter or an underscore — never a digit.
- Names are case-sensitive: `age`, `Age` and `AGE` are three different variables.
- Python **keywords**, such as `if`, `for`, `while`, `True` and `def`, cannot be used as names. (Run `import keyword; print(keyword.kwlist)` to see the full list.)

**Good practice** (not enforced, but strongly recommended):

- Choose meaningful names: `my_age`, `initial_speed`, `profit` rather than `a`, `b`, `c`.
- Avoid names that are easily confused, such as `l` (lower-case L), `O` (upper-case o) and `I` (upper-case i) — unless your aim is to confuse!
- Avoid the names of built-in functions, such as `print`, `sum`, `max`, `min`, `list` and `str`. Python allows it, but the original function is then hidden ("shadowed"), and you can no longer use it.

There are two popular conventions for names made of several words:

1. **Camel case**: `myAge`, `initialSpeed`, `myProfit`
2. **Snake case** (the underscore convention): `my_age`, `initial_speed`, `my_profit`

Whichever you choose, be consistent. Python's official style guide, [PEP 8](https://peps.python.org/pep-0008/), recommends snake case for variables and functions, and capitalised words (e.g. `ElectricCar`) for class names, which we meet in Lecture 5. **Throughout this course, we use snake case.** Some older languages restricted the length of names, which encouraged short and cryptic names; modern languages have no such restriction, and underscores make long names much easier to read.

> **💡 Long names cost nothing**
>
> There is no need to save keystrokes on variable names. Modern editors and IDEs (Jupyter, VS Code, PyCharm, Spyder, …) complete names for you: type the first few letters and press **Tab**.

## Creating and updating variables

**Creating a variable.** A variable is created by an **assignment**: the value on the right-hand side (RHS) of the equals sign `=` is computed first, and the result is then bound to the name on the left-hand side (LHS). The RHS may be a literal, another variable, or a complex expression involving several variables.

For example, in the statement `i = 7`, the RHS is evaluated and the result is assigned to `i`. We may read this as "`i` becomes equal to 7", or "`i` holds the value 7". In the statement

```python
result = (result_1 + result_2 / result_3) * result_4
```

Python first evaluates the RHS using the values currently held by `result_1`, `result_2`, `result_3` and `result_4`, and then assigns the outcome to `result`. In both cases, Python brings the variable into existence at the moment it is first assigned; there is no need to declare it in advance.

**Updating a variable.** To update a variable, we simply assign a new value to the existing name.

````python
# 'Albert' is a literal; first_name is a variable (single quotes)
first_name = 'Albert'
last_name = "Einstein"  # double quotes work just as well
# evaluate the RHS, then assign it to the LHS
full_name = first_name + ' ' + last_name
````

````python
# The variables keep their values, so we can use them at any time.
# print() writes whatever we give it to the output area.
print("Hello world")   # apply a function to a literal
print(first_name)      # apply a function to a variable
print(last_name)
print(full_name)
````

````python
# Some more (rather useless) code, just for fun
my_look_age = 25                    # my_look_age becomes equal to 25
my_real_age = 35                    # my_real_age becomes equal to 35
ratio = my_look_age / my_real_age   # division, just like at school
print(ratio)
````

````python
# Updating a variable: the right-hand side uses the OLD value
speed = 10
print(speed)
speed = speed + 5    # read as: "speed becomes the old speed plus 5"
print(speed)
````

# Basic data types

Every value in Python has a **type**, which determines what the value represents and which operations are allowed on it. Python works out the type automatically from the value; you never need to declare it. The most common basic types are:

| Type       | Meaning                          | Examples                         |
|------------|----------------------------------|----------------------------------|
| `int`      | whole numbers (of any size)      | `7`, `-42`, `10**100`            |
| `float`    | floating-point (real) numbers    | `3.14`, `-0.5`, `2.0`, `1.5e-3`  |
| `complex`  | complex numbers                  | `3 + 4j`                         |
| `bool`     | Boolean (truth) values           | `True`, `False`                  |
| `str`      | text (strings)                   | `'spam'`, `"Hello World"`        |
| `NoneType` | "no value at all"                | `None`                           |

The built-in function `type()` tells us the type of any value.

````python
print(type(7))
print(type(3.14))
print(type(3 + 4j))
print(type(True))
print(type("Hello"))
print(type(None))
````

## Converting between types

The names of the types can also be used as functions that convert a value into that type. This is particularly important when we read input from the keyboard, which always arrives as a string (see the section on input below).

````python
print(int("42") + 1)     # string -> integer
print(float("3.5") * 2)  # string -> float
print(str(2024) + "!")   # integer -> string
# float -> integer: the decimal part is cut off (not rounded!)
print(int(3.99))
print(round(3.99))       # use round() if you want rounding
print(bool(0), bool(5))  # 0 converts to False; any other number converts to True
````

A conversion that does not make sense raises an error:

````python
int("3.5")   # ValueError: "3.5" is not the text of an integer
````

## Everything is an object

We will study **classes** and **objects** properly in Lecture 5. Roughly speaking, an *object* is an entity that has its own data, features and actions, and a *class* is the template from which such objects are made.

For example, "Mary" is a particular object of the template "Student at the University of Glasgow". Mary has a name, a GUID and a list of registered courses and, like every other student, she has to enrol on courses and pass exams in order to graduate. Likewise, "Iron Man" is a particular object of the template "Superhero": he has a real name and special powers, and he can do fancy things, just like every other superhero — although each hero has their own name, powers and tricks.

In Python:

- Every value — integers, floating-point numbers, strings, and so on — is actually an object, and a variable is simply a name bound to an object.
- Objects carry useful information with them. For instance, a string knows its own length, so Python never has to count its characters again.
- Because a variable is only a name, we can remove the name with `del`. Afterwards, the name is undefined.

````python
x = 42
print(x)
del x       # remove the name x
print(x)    # NameError: name 'x' is not defined
````

# Operators

In simple terms, **operators** are symbols that act on values or variables to produce new values — mostly maths, just like at school. Addition, subtraction, multiplication and division are all operators acting on two numbers (integers or floating-point numbers). In more technical language, **operators** are constructs defined within a programming language that perform operations on values and variables, which are called **operands**.

> **❗ Take-home message**
>
> - Every program is a sequence of modifications of variables — that is what an algorithm is about.
> - These modifications happen through operators, e.g. plus `+`, minus `-`, multiply `*`, divide `/`, etc.
> - Python's built-in libraries provide more advanced operators, shortcuts and functions. We will learn them over time.

## Types of operators

Python supports the following types of operators:

- arithmetic operators;
- comparison (relational) operators;
- logical operators;
- membership operators;
- identity operators;
- assignment operators;
- bitwise operators.

We study all of them except the bitwise operators, which act on the individual bits of integers and are rarely needed in this course.

## Arithmetic operators

Assuming `a = 10` and `b = 20`:

| Operator | Description                                | Example                                   |
|----------|--------------------------------------------|-------------------------------------------|
| `+`      | addition                                   | `a + b` gives `30`                        |
| `-`      | subtraction                                | `a - b` gives `-10`                       |
| `*`      | multiplication                             | `a * b` gives `200`                       |
| `/`      | division                                   | `a / b` gives `0.5`                       |
| `%`      | modulus (remainder of a division)          | `a % b` gives `10`                        |
| `**`     | exponentiation (power)                     | `2 ** 4` gives $2^4 = 16$                 |
| `//`     | floor division (round *down* the quotient) | `9 // 2` gives `4`; `-11 // 3` gives `-4` |

Note that the division `/` always returns a `float`, even when the result is a whole number.

````python
a, b, c, d, e, f, g = 5, 1, 2, 3, 4, 7, 8
# This is a short way to assign several values to several variables at once:
# a = 5, b = 1, c = 2, d = 3, e = 4, f = 7, g = 8
print(g // f)
print(c ** d)
print(a + b * c * d % e - f / g)
# the same expression, with brackets showing the order
print(a + (b * c * d) % e - (f / g))
print(4 / 2)  # / always returns a float
````

## Comparison operators

Comparison operators compare two values and return a Boolean value: `True` or `False`.

| Operator | Description              |
|----------|--------------------------|
| `<`      | less than                |
| `<=`     | less than or equal to    |
| `>`      | greater than             |
| `>=`     | greater than or equal to |
| `==`     | equal to                 |
| `!=`     | **not** equal to         |

> **⚠️ `=` is not `==`**
>
> A single equals sign `=` *assigns* a value to a variable; a double equals sign `==` *compares* two values.

````python
a, b, c, d = 1, 2, 3, 3
print("a < b  :", a < b)
print("a > b  :", a > b)
print("c <= d :", c <= d)
print("c >= d :", c >= d)
print("a == b :", a == b)
print("a != b :", a != b)
````

## Logical operators

Boolean values can be combined with logical operators. We rarely store Boolean values explicitly in variables; they usually come from comparisons, as in the last example of the next cell.

| Operator | Description                                              |
|----------|----------------------------------------------------------|
| `not`    | negation: `not x` is `True` if `x` is `False`            |
| `and`    | `x and y` is `True` only if **both** `x` and `y` are `True` |
| `or`     | `x or y` is `True` if **at least one** of them is `True` |

````python
print("True and False =", True and False)
print("False or True  =", False or True)
print("not False      =", not False)
print("not True       =", not True)

temperature = 18
print("Pleasant weather:", temperature > 15 and temperature < 25)
````

## Membership operators

Membership operators test whether a value occurs in a sequence. We use them with strings for now, and with lists, tuples and dictionaries in Lecture 3.

| Operator | Description                                          |
|----------|------------------------------------------------------|
| `in`     | `True` if the value is found in the sequence         |
| `not in` | `True` if the value is **not** found in the sequence |

````python
word = "engineering"
print("gin" in word)
print("z" in word)
print("z" not in word)
````

## Identity operators

Identity operators test whether two names refer to the **same object** (recall the figure in the section *Variables in a nutshell*).

| Operator | Description                                  |
|----------|----------------------------------------------|
| `is`     | `True` if both refer to the same object      |
| `is not` | `True` if they refer to **different** objects |

````python
x = 1
y = x
z = 3
# y refers to the same integer object as x, so their IDs are identical.
# z refers to a different object, so its ID is different.
print("id(x) =", id(x))
print("id(y) =", id(y))
print("id(z) =", id(z))
print("x is y     :", x is y)
print("x is not y :", x is not y)
print("x is z     :", x is z)
print("x is not z :", x is not z)
````

> **⚠️ `is` versus `==`**
>
> `==` checks whether two values are *equal*; `is` checks whether two names refer to the *same object*. Two objects can be equal without being the same object, as the next cell shows (it uses lists, which we study in Lecture 3). Use `==` to compare values, and keep `is` for special cases such as `x is None`.

````python
p = [1, 2, 3]
q = [1, 2, 3]
print(p == q)   # True: the two lists have equal contents
print(p is q)   # False: they are two different objects
````

## Assignment operators

- The equals sign `=` is the **assignment operator**: it assigns the value of the right-hand side to the variable on the left-hand side. So `x = 3 + 5` is fine, but `8 = 3 + 5` is a syntax error, because nothing can be assigned to the literal `8`.
- Statements of the form `n = n + 1` are so common that Python offers a shortcut: `n += 1`. Note that the equals sign comes *after* the arithmetic operator. The same shortcut exists for all the arithmetic operators.

| Operator | Example   | Equivalent to          |
|----------|-----------|------------------------|
| `=`      | `x = 3`   | `x` becomes equal to 3 |
| `+=`     | `x += 3`  | `x = x + 3`            |
| `-=`     | `x -= 3`  | `x = x - 3`            |
| `*=`     | `x *= 3`  | `x = x * 3`            |
| `/=`     | `x /= 3`  | `x = x / 3`            |
| `%=`     | `x %= 3`  | `x = x % 3`            |
| `//=`    | `x //= 3` | `x = x // 3`           |
| `**=`    | `x **= 3` | `x = x ** 3`           |

````python
n = 10
n += 1     # n = n + 1
print(n)
n *= 3     # n = n * 3
print(n)
n //= 4    # n = n // 4
print(n)
````

## Operator precedence and associativity

- **Precedence** works as in ordinary mathematics: `**` binds more tightly than `*` and `/`, which in turn bind more tightly than `+` and `-`.
- When operators have the same precedence, **associativity** decides the order of evaluation. Most operators are evaluated from **left to right**; for example, `10 - 4 - 3` means `(10 - 4) - 3 = 3`.
- Exponentiation `**` is the exception: it is evaluated from **right to left**, so `2 ** 3 ** 2` means `2 ** (3 ** 2) = 512`.

**Always use brackets if you are even vaguely unsure.** They cost nothing and make your intention clear. The complete precedence table is in the [official Python documentation](https://docs.python.org/3/reference/expressions.html#operator-precedence).

````python
print(10 - 4 - 3)     # left to right: (10 - 4) - 3
print(2 ** 3 ** 2)    # right to left: 2 ** (3 ** 2)
print((2 ** 3) ** 2)  # brackets change the order
# careful: ** binds more tightly than the minus sign, so this is -(2 ** 2)
print(-2 ** 2)
````

# Strings

Strings are among the most natural data types we use every day, alongside integers and floating-point numbers. A string is a sequence of characters, so it can hold a single letter, a word or even a whole paragraph.

## Creating strings

Strings can be enclosed in single quotes (`'...'`) or double quotes (`"..."`) with the same result. The backslash `\` is used to **escape** characters that would otherwise have a special meaning: to put a single quote inside a string enclosed in single quotes, write `\'`; similarly, `\"` puts a double quote inside a string enclosed in double quotes. The most useful escape sequences are:

| Escape sequence | Meaning              |
|-----------------|----------------------|
| `\n`            | new line             |
| `\t`            | tab                  |
| `\'`            | single quote         |
| `\"`            | double quote         |
| `\\`            | a backslash itself   |

````python
string_1 = 'spam eggs'                    # single quotes
string_2 = 'It doesn\'t mean food.'       # use \' to escape the single quote
# Inside double quotes, a single quote needs no escape character.
string_3 = "It doesn't mean food."
# This makes sense: if a string started with a single quote, how could Python
# tell whether the next single quote ends the string or belongs to it?

# The character "\n" starts a new line, so this string is printed on two lines.
string_4 = "This is the first line.\nThis is the second one."
# The character "\t" inserts a tab, i.e. some horizontal space, like the TAB key.
string_5 = "I create a tab \t and you see so much space.\t\t What, not again!"
string_6 = "Compare a single space ' ' with a tab '\t'."

print(string_1)
print(string_2)
print(string_3)
print(string_4)
print(string_5)
print(string_6)
````

Triple quotes create a string that spans several lines:

````python
poem = """Roses are red,
violets are blue,
Python is great,
and so are you."""
print(poem)
````

## Operations on strings

- Strings can be concatenated (glued together) with the `+` operator, and repeated with the `*` operator.
- Two or more string *literals* (i.e. text enclosed between quotes) written next to each other are concatenated automatically. This is particularly useful for breaking long strings over several lines.
- The built-in function `len()` returns the number of characters in a string.
- Individual characters are accessed by their position (counting from zero), e.g. `word[0]`. We study indexing in detail with lists in Lecture 3.

````python
a = "You"
b = "are"
c = "awesome"
# strings stored in variables, concatenated
print(a + " " + b + " " + c + "!")
print("I" + " am" + " Coffee " + "Milk.")   # with plus signs
print("I" " am" " Coffee " "Milk.")         # without plus signs: adjacent literals
print(3 * "ha")                             # repetition
print(len("engineering"))                   # number of characters
print("engineering"[0], "engineering"[-1])  # first and last character
````

> **❗ Strings are immutable**
>
> Once a string has been created, it cannot be changed. Operations that seem to modify a string actually create a **new** string.

````python
greeting = "Hello"
greeting[0] = "J"    # TypeError: strings do not support item assignment
````

## String methods

Every string is an object, so it comes with many useful **methods** — functions attached to the object — called with the dot syntax `my_string.method_name()`. Here are a few of them:

````python
s = "  Python for Engineers  "
print(s.upper())                      # all upper case
print(s.lower())                      # all lower case
print(s.strip())                      # remove leading and trailing spaces
print(s.replace("Python", "Maths"))   # returns a NEW string
print("Hello".replace("H", "J"))      # the way to "change" a string
print("12345".isnumeric(), "a1b2".isnumeric())
````

# Output with `print()`

Producing neat output was historically one of the messiest parts of many programming languages; output conventions date from a time when dot-matrix line printers were cool! Fortunately, Python makes it fairly painless. We discuss output first and input afterwards.

## The basics

- `print()` accepts several values separated by commas. By default, it separates them with a single space; the keyword argument `sep` changes the separator.
- `print()` always ends with an invisible newline character `\n`. You can change this with the keyword argument `end`; for example, `print('something', end='')` prints nothing after its argument, so the next output continues on the same line.

````python
print("Something")
print("Wrong")                   # a new line starts after "Wrong"

print("-------------------")
print("Something", end="")       # NO new line
print("Else", end=",\n")         # a comma and then a new line after "Else"

print("-------------------")
print("x", "y", "z")             # separated by spaces (the default)
print("x", "y", "z", sep=" | ")  # a custom separator
````

## Formatting output with `str.format()`

Because strings are objects, they come with the method `format()`, which is used to produce fancy output. Best explained by an example:

```python
print("My text {} {} {}.".format(arg1, arg2, arg3))
```

prints `My text` and then substitutes the values of the three arguments `arg1`, `arg2` and `arg3` into the three placeholders `{}`, in order of appearance. For example, if `arg1 = 'is'`, `arg2 = 'very'` and `arg3 = 'useful'`, the output is `My text is very useful.`

Each placeholder can contain a specification of the form `{<select>:<format>}`. Here, `<select>` is the position of the argument in the argument list (counting from zero), so the following two lines are identical:

```python
print("a = {}, b = {}".format(arg1, arg2))
print("a = {0}, b = {1}".format(arg1, arg2))
```

The `<format>` part, after the colon, controls the alignment, width, precision and so on. The most useful specifications are:

| Specification | Meaning                                         | Example result for `3.14159` / `'hi'` |
|---------------|-------------------------------------------------|---------------------------------------|
| `{:<10}`      | left-align in a field 10 characters wide        | `'hi        '`                        |
| `{:>10}`      | right-align in a field 10 characters wide       | `'        hi'`                        |
| `{:^10}`      | centre in a field 10 characters wide            | `'    hi    '`                        |
| `{:*^10}`     | centre, filling the space with `*`              | `'****hi****'`                        |
| `{:.3f}`      | fixed-point notation with 3 decimal places      | `'3.142'`                             |
| `{:10.3f}`    | width 10, 3 decimal places                      | `'     3.142'`                        |
| `{:.2e}`      | scientific notation with 2 decimal places       | `'3.14e+00'`                          |
| `{:+f}`       | always show the sign                            | `'+3.141590'`                         |

The following examples are self-explanatory:

````python
print("{:<40}".format('left aligned'))              # < : aligned to the left
print("{:>40}".format('right aligned'))             # > : aligned to the right
print("{:^40}".format('centred'))                   # ^ : centred
print("{:*^40}".format('centred with asterisks'))   # fill with * and then centre
````

````python
print("{0:+f}; {1:+f}".format(3.14, -3.14))   # always show the sign
print("{0: f}; {1: f}".format(3.14, -3.14))   # a space instead of the plus sign
print("{0:-f}; {1:-f}".format(3.14, -3.14))   # only the minus sign (the default)
print("{0:.2f} and {0:.2e}".format(1234.5678))  # the same argument used twice
````

## f-strings: the modern way

Since Python 3.6, there has been an even more convenient way of formatting: **formatted string literals**, or **f-strings**. Put the letter `f` before the opening quote and write variables — or any expression — directly inside the braces. The format specifications after the colon are exactly the same as above. f-strings are the recommended style for new code, but you will meet both styles in these notes and elsewhere.

````python
name = "Ada"
speed = 12.3456
print(f"{name} cycles at {speed:.2f} m/s.")
print(f"{'centred':*^20}")
print(f"2 + 3 = {2 + 3}")
````

# A first look at modules

Python comes with a large **standard library** organised into **modules**, and there are thousands of third-party packages such as NumPy. To use the functions in a module, we first `import` it. There are three common forms of the import statement:

````python
import math                 # import the whole module ...
print(math.sqrt(16))        # ... and use its functions with the prefix "math."
print(math.pi)

from math import sqrt       # import a single function ...
print(sqrt(2))              # ... and use it without the prefix

import numpy as np          # import a module under a shorter alias
print(np.cos(np.pi))
````

# Input with `input()`

- Signature: `input(prompt='')`. The `prompt` is displayed, and Python waits until the user types something and presses **Enter**.
- Typical use: `s = input("Question? ")`.
- `input()` **always returns a string**, even if the user types a number. Therefore, we usually need a conversion function to obtain the number we want, e.g. `my_input = int(s)` or `my_input = float(s)`.

## Example: the roots of a quadratic equation

We ask the user for the three coefficients of the quadratic equation $c_2 x^2 + c_1 x + c_0 = 0$ and then compute its two roots,

$$
x_{1,2} = \frac{-c_1 \pm \sqrt{\Delta}}{2 c_2}, \qquad \text{where } \Delta = c_1^2 - 4 c_2 c_0
$$

is the **discriminant**.

````python
from math import sqrt

c0 = float(input("Coefficient of x^0: "))  # input() returns a string, so convert it
c1 = float(input("Coefficient of x^1: "))
c2 = float(input("Coefficient of x^2: "))

discriminant = c1**2 - 4*c2*c0   # value**2 means value to the power of 2
denominator = 2*c2
sqrt_discriminant = sqrt(discriminant)

root_1 = (-c1 + sqrt_discriminant) / denominator
root_2 = (-c1 - sqrt_discriminant) / denominator
print("The two roots are:\n r1 = {0}\n r2 = {1}".format(root_1, root_2))
````

A sample run for the equation $x^2 - 3x + 2 = 0$ looks like this:

```text
Coefficient of x^0: 2
Coefficient of x^1: -3
Coefficient of x^2: 1
The two roots are:
 r1 = 2.0
 r2 = 1.0
```

> **⚠️ What if Δ < 0?**
>
> If the discriminant is negative, the equation has no real roots, and `sqrt()` stops the program with `ValueError: math domain error`. To deal with this case, the program has to make a *decision* — exactly what the `if` statement does. We will revisit this example in Lecture 2.

# Summary

- A **literal** is a fixed value in the code; a **variable** is a name that refers to an object in memory. Variables are created by assignment, updated by reassignment and removed with `del`.
- Follow the naming rules, choose meaningful names, use snake case, and never shadow built-in names such as `sum` or `max`.
- The basic data types are `int`, `float`, `complex`, `bool`, `str` and `NoneType`. Use `type()` to inspect a value and `int()`, `float()`, `str()` to convert between types.
- Operators come in families — arithmetic, comparison, logical, membership, identity and assignment — and follow precedence and associativity rules. When in doubt, use brackets.
- Strings are immutable sequences of characters with many useful methods. Use `print()` with `str.format()` or, preferably, f-strings for tidy output.
- Use `import` to access modules such as `math` and `numpy`.
- `input()` always returns a string; convert it before doing arithmetic.
