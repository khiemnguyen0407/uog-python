```python
import matplotlib.pyplot as plt

with plt.xkcd(scale=0.8, length=100, randomness=2):
    fig, ax = plt.subplots(figsize=(10, 5))

    ax.text(
        0.2,
        0.7,
        "a + b",
        ha="center",
        va="center",
        fontsize=18,
        bbox={
            "boxstyle": "round,pad=0.4",
            "facecolor": "white",
            "edgecolor": "#17204f",
            "linewidth": 2,
        },
    )

    ax.annotate(
        "",
        xy=(0.65, 0.7),
        xytext=(0.35, 0.7),
        arrowprops={
            "arrowstyle": "->",
            "linewidth": 2,
            "color": "#17204f",
        },
    )

    ax.text(
        0.8,
        0.7,
        "a.__add__(b)",
        ha="center",
        va="center",
        fontsize=18,
        bbox={
            "boxstyle": "round,pad=0.4",
            "facecolor": "#dce8ff",
            "edgecolor": "#17204f",
            "linewidth": 2,
        },
    )

    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis("off")

    plt.savefig("method_call_diagram.png", dpi=300, bbox_inches="tight")
```
